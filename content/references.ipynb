## Build and browser runtime

- Jupyter Book contributors. [Jupyter Book v1 documentation](https://jupyter-book.readthedocs.io/v1/intro.html). Stable v1/Sphinx renderer, in maintenance mode; this project pins its tested build stack.
- JupyterLite contributors. [Documentation](https://jupyterlite.readthedocs.io/en/stable/), [content and storage](https://jupyterlite.readthedocs.io/en/stable/howto/content/files.html), [deployment](https://jupyterlite.readthedocs.io/en/stable/howto/deployment/index.html).
- JupyterLite contributors. [Configuration files](https://jupyterlite.readthedocs.io/en/stable/howto/configure/config_files.html) and [advanced extension configuration](https://jupyterlite.readthedocs.io/en/stable/howto/configure/advanced/extensions.html). The atlas uses an isolated Lite project, explicit distribution-owned Pyodide extension assets, and a forced clean assembly.
- Pyodide contributors. [Pyodide documentation](https://pyodide.org/en/stable/) and [package loading](https://pyodide.org/en/stable/usage/loading-packages.html). Python compiled to WebAssembly; initial downloads need network.
- GitHub. [Custom workflows with GitHub Pages](https://docs.github.com/en/pages/getting-started-with-github-pages/using-custom-workflows-with-github-pages).
- GitHub Actions. [setup-python v5: dependency caching](https://github.com/actions/setup-python/blob/v5/README.md#caching-packages-dependencies). The workflow explicitly keys its cache from both pinned requirements files.
- STAC community. [STAC specification overview](https://stacspec.org/en/about/stac-spec/). Items, collections, catalogs, assets and rights metadata are distinct.

## Human-centered experimentation and accountable AI

- NIST/SEMATECH. [e-Handbook of Statistical Methods: Linear Least Squares Regression](https://www.itl.nist.gov/div898/handbook/pmd/section1/pmd141.htm). Method background and limitations for the original fitted-model foundation.
- Scikit-learn developers. [Cross-validation: evaluating estimator performance](https://scikit-learn.org/stable/modules/cross_validation.html#cross-validation-iterators-for-grouped-data). Held-out evaluation and grouped data; referenced for teaching, not installed by the notebook.

- IDEO.org. [Design Kit methods](https://www.designkit.org/methods.html). Participatory inquiry, practical prototyping and feedback inform the atlas's original listen/frame/prototype/learn diagram.
- World Health Organization (2021). [Ethics and governance of artificial intelligence for health](https://www.who.int/publications/i/item/9789240029200). Human rights, accountability and community responsiveness inform the evidence-to-action synthesis.
- de Bono, Edward. [Six Thinking Hats](https://www.debono.com/Books/six-thinking-hats). Framework attributed to its creator. Purple/orchestrator additions belong to the linked projects and are not one of the original six.
- Zuckerman, Ethan (2013). [Rewire: Digital Cosmopolitans in the Age of Connection](https://ethanzuckerman.com/books/rewire/). Author's book page. The Rewiring source course connects cross-cultural understanding to its health learning approach.
- Tobias, James L. (2013). *Rewiring to meet the challenges of "Wicked Problems" within global and domestic Public Health*. Spatial Plexus presentation, cited through the [Rewiring source README](https://github.com/jltobias/JupyterLite-Rewiring-Global-Health/blob/main/README.md); original presentation not independently inspected for this atlas.
- GISGeography. [Periodic Table for Spatial Analysis](https://gisgeography.com/spatial-analysis-periodic-table/) and [citation guidance](https://gisgeography.com/how-to-cite/). Inspiration for linked spatial-method collections, not copied artwork.

## Domain source context (linked, not bundled data)

- OpenGeos contributors. [RustGIS source](https://github.com/opengeos/RustGIS), [official documentation](https://rustgis.app/) and Apache 2.0 license. Cite the version/commit used; native Python bindings are distinct from RustGIS Web and the source course's MapLibre/Three.js companion viewers.
- JupyterGIS contributors / QuantStack. [JupyterGIS documentation](https://jupytergis.readthedocs.io/en/latest/) and [story-map guide](https://jupytergis.readthedocs.io/en/latest/user_guide/how-tos/story-maps.html). Source bundles and browser previews have separate environment requirements.

- Wu, Qiusheng (2026). *GeoLibre: A lightweight, cloud-native GIS platform for visualizing, exploring, and analyzing geospatial data*. [Zenodo DOI](https://doi.org/10.5281/zenodo.20785400), cited as provided by the GeoLibre course README.
- Wu, Qiusheng (2026). *GeoAI: A Python package for integrating artificial intelligence with geospatial data analysis and visualization*. JOSS 11(118), 9605. [DOI](https://doi.org/10.21105/joss.09605), cited as provided by the source README.
- Rhee S-Y, Blanco JL, Jordan MR, et al. (2015). *Geographic and Temporal Trends in the Molecular Epidemiology and Genetic Mechanisms of Transmitted HIV-1 Drug Resistance*. PLOS Medicine 12(4), e1001810. [DOI](https://doi.org/10.1371/journal.pmed.1001810). Historical source of the HIVDB course baseline, not a current-surveillance claim.
- Tobias, James L., Maria Jolly, Kelly Van Baalen, Leila Hadj-Chikh, Edward Paul Vallejo, Nathan Heard, Zhanar Haimovich, Casey Kalman, Baboyma Kagniniwa, Jesse Blanton and Herman Tolentino (2024). *PEPFAR Adapts to Sea Level Rise and Storm-Surge: Potential Impacts to Coastal Cities and Facilities (2030)*. AIDS 2024 poster THPEF698. [Official IAS poster](https://plus.iasociety.org/sites/default/files/2024-09/e-poster_755.pdf), referenced through the Sea-Level-Rise course; poster image not copied.
- European Commission JRC. [GHSL Urban Centre Database R2024A](https://human-settlement.emergency.copernicus.eu/ghs_ucdb_2024.php). Urban-centre data are not informal-settlement classifications.
- [dzole0311/zarr-sql-views](https://github.com/dzole0311/zarr-sql-views), upstream inspiration credited by the Zarr/SQL teaching atlas; upstream ISC terms remain separate.
- Original companion inspirations: [God's Eye View by Bilawal Sidhu](https://github.com/bilawalsidhu/gods-eye-view), [World View by Kevin (Khoa) To](https://github.com/kevtoe/worldview), [OSIRIS](https://github.com/simplifaisoul/osiris), [World Monitor](https://www.worldmonitor.app/), and [Awesome GPT-6 Astra](https://github.com/magiccreator-ai/awesome-gpt-6-astra). Linked source courses provide their specific attribution and limits; this atlas imports none of their code or data.
