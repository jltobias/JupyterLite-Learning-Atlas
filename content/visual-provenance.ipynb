# Original atlas visuals

All six SVGs are original vector compositions by James L. Tobias and JupyterLite Learning Atlas contributors (2026), generated from `scripts/make_assets.py` and licensed CC BY 4.0. No stock images, source screenshots, third-party infographics, logos, geographic boundaries, or remote fonts are embedded.

| File | Meaning and provenance |
|---|---|
| `atlas-mark.svg` | Original globe/open-notebook symbol; no third-party logo. |
| `splash.svg` | Abstract globe, notebook, layer stack and shared learning. Geography is illustrative, not a dataset. |
| `capabilities.svg` | Original explanation of browser activities and separate native/GPU/service continuation paths. |
| `evidence-to-action.svg` | Original evidence/review/action learning loop; conceptually informed by WHO AI governance and the Rewiring source course. |
| `empathy-iteration.svg` | Original listen/frame/prototype/learn diagram; conceptually informed by IDEO.org Design Kit. |
| `repository-overview.svg` | Actual counts of 34 projects (31 JupyterLite-named plus 3 related) across seven themes from `catalog/repositories.json`, not synthetic observations. Regenerate after catalog changes. |

Each file has a title, description and ARIA labeling. Notebook-generated SVG figures are original deterministic drawings of their labeled synthetic data (or explicit catalog metadata); every chart prints values as a text alternative. The interactive teaching views use original projected SVG geometry and a fictional 4×4 grid. No actual country, settlement, facility or person is represented.

Copies in `content/assets/` and assembled `_site/assets/` / `_site/lite/files/assets/` come from these exact assets. Attribution: “James L. Tobias and contributors, JupyterLite Learning Atlas (2026), CC BY 4.0.” See the root `REFERENCES.md` for conceptual sources.

The four newly public source companions add original threshold-count, focal-mean, inclusive planar-buffer and uneven-extract figures. Their values and geometry are invented, printed as text, and use no external maps, source datasets or source screenshots. The Shiny-stub figure shows an original rule prototype, not a working source application.

Ocean Watch adds an original AIS–SAR candidate-count SVG from invented planar km/minute positions; it retains ambiguous and unmatched categories. Zarr/SQL adds an original per-variable covered-cell-day SVG from invented 8/5-day intervals and None masks. Both print their current values, units and denominators. No source course artwork, satellite image, data or LexCube renderer is embedded. The repository overview counted 33 chapters at the Ocean/Earth refresh; all other compositions are preserved.

The targeted Coral Reef One Health addition (2026-11-02) adds an original SVG with fictional two-descriptor indices and observed weight coverage. Unknown is drawn explicitly, never as zero; the title, printed results and figure follow edited weights and missingness. No source splash, reef map, photograph, logo, dataset or community observation is copied. Only the catalog-derived `repository-overview.svg` is refreshed to 34 chapters; other original artwork compositions are preserved.
