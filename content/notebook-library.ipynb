# The source notebook library

Read full pinned source lessons alongside the original short Atlas companions. Each notebook is the same canonical file in Book and Lab. Runtime labels describe intended requirements; execution receipts identify the smaller set actually tested. First runtime/package downloads require connectivity. Download important edits and exports from browser storage.

**Browser local:** bundled calculation inputs. **Browser network:** remote APIs, tiles, packages or services. **Local/server:** native GIS, shell, legacy or installed services. **Reference only:** missing/conditional inputs. A local Lab link is an editing/reading handoff, not a promise every branch executes there.


## Scope and provenance

46 pinned source repositories; 825 inspected notebook paths; 777 distinct authored lessons, 19 inherited compatibility examples, 12 exact aliases and 17 rights-excluded legacy notebooks. Source code and prose are preserved except documented format, provenance, link, rights and browser compatibility adaptations. Unverified upstream outputs were cleared. Scoped notices and file hashes are recorded in the [published source manifest](https://github.com/jltobias/JupyterLite-Learning-Atlas/blob/main/catalog/source-library.json).

Four spatial periodic-table variants are retained as separate course editions: synthetic methods, map-centric, older helper and current map/helper editions. The current map/helper guide gives the richest local map/data context; matching titles do not establish duplicates.
