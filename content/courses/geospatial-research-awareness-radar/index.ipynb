## Browser GIS pathway

**Geographic question:** Which works have a verified study area?

**Data readiness (N + optional O):** deduplication companion, morning-scan source, and sample metadata.

**Start from the chapter evidence:** [deduplication companion](../../projects/geospatial-research-awareness-radar/index.ipynb) · [morning-scan source](01_morning_scan.ipynb) · [sample metadata](https://github.com/jltobias/JupyterLite-Learning-Atlas/blob/main/content/courses/geospatial-research-awareness-radar/sample_results_2026-09-04.json) (`courses/geospatial-research-awareness-radar/sample_results_2026-09-04.json`)

- **[RustGIS Web](https://web.rustgis.app/):** Deduplicate sample metadata; attach study_area_feature_id only after human verification against a selected layer. Keep unknown areas in the table. In RustGIS, import the reviewed geographic export through Add Data; inspect the selected feature IDs and attribute table before styling.
- **[GeoLibre Web](https://web.geolibre.app/):** Deduplicate sample metadata; attach study_area_feature_id only after human verification against a selected layer. Keep unknown areas in the table. In GeoLibre, import that same reviewed file, compare a second legend/filter, and reconcile its count and one known attribute with the notebook.

**Handoff:** Use the [GIS workbench](../../gis-workbench.ipynb) to validate/export an attributed WGS84 layer. Download the GeoJSON, method record and review prompt from `exports/`, then import through the chosen app’s file picker. Compare `_atlas_source_id` (or source `GEOID`) with the original IDs; app-generated internal IDs may differ. Return changed results by explicit upload. Apps and the notebook share no files or kernel state. Practice-only Census data remain Fresno 2020, never this subject’s observations.

**Optional AI prompt (approved subject evidence only):** “Cite the supplied record supporting each geography; abstain otherwise. Cite supplied feature IDs/fields, separate observations from assumptions, abstain when evidence is missing, and propose one falsifiable check.” For the common practice package, use its generated schema/summary prompt instead. A human-written answer completes the same task.

**GeoAI opportunity:** Spatial literature coverage is an evidence audit, not hazard inference. Any later fitted exercise must name target/features and spatial/temporal support, reserve a spatial/temporal holdout, inspect leakage and uncertainty, and state evidence that would disprove the interpretation. Styling, thresholds, schema checks and chat alone are not model fitting.

**Human evidence check:** No title/author-affiliation geocoding; distinct DOI counts reconcile. Ask the intended reader which decision the map supports; change one legend, filter or annotation, then recheck interpretation against the same IDs and values.
