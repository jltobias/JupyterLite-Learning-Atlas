## Browser GIS pathway

**Geographic question:** How can symbol choice misrepresent the same events?

**Data readiness (B):** USGS scene lesson, event table, and manifest. The CSV contains longitude, latitude, depth in kilometres, UTC time and magnitude; its companion GeoJSON has third ordinates.

**Start from the chapter evidence:** [USGS scene lesson](repository_examples/07_usgs_earthquake_scenes.ipynb) · [event table](https://github.com/jltobias/JupyterLite-Learning-Atlas/blob/main/content/courses/data-visualization/repository_examples/data/usgs/events.csv) (`courses/data-visualization/repository_examples/data/usgs/events.csv`) · [manifest](https://github.com/jltobias/JupyterLite-Learning-Atlas/blob/main/content/courses/data-visualization/repository_examples/data/manifest.json) (`courses/data-visualization/repository_examples/data/manifest.json`)

- **[RustGIS Web](https://web.rustgis.app/):** Export a small dated subset of the bundled USGS CSV as 2D points; retain depth_km. Compare legends and inspect IDs/units. In RustGIS, import the reviewed geographic export through Add Data; inspect the selected feature IDs and attribute table before styling.
- **[GeoLibre Web](https://web.geolibre.app/):** Export a small dated subset of the bundled USGS CSV as 2D points; retain depth_km. Compare legends and inspect IDs/units. In GeoLibre, import that same reviewed file, compare a second legend/filter, and reconcile its count and one known attribute with the notebook.

**Handoff:** Use the [GIS workbench](../../gis-workbench.ipynb) to validate/export an attributed WGS84 layer. Download the GeoJSON, method record and review prompt from `exports/`, then import through the chosen app’s file picker. Compare `_atlas_source_id` (or source `GEOID`) with the original IDs; app-generated internal IDs may differ. Return changed results by explicit upload. Apps and the notebook share no files or kernel state. Practice-only Census data remain Fresno 2020, never this subject’s observations.

**Optional AI prompt (approved subject evidence only):** “Critique a misleading magnitude legend using actual IDs. Cite supplied feature IDs/fields, separate observations from assumptions, abstain when evidence is missing, and propose one falsifiable check.” For the common practice package, use its generated schema/summary prompt instead. A human-written answer completes the same task.

**GeoAI opportunity:** Optional existing DBSCAN comparison; magnitude display is not fitting. Any later fitted exercise must name target/features and spatial/temporal support, reserve a spatial/temporal holdout, inspect leakage and uncertainty, and state evidence that would disprove the interpretation. Styling, thresholds, schema checks and chat alone are not model fitting.

**Human evidence check:** Count, ID, magnitude, time and depth survive transfer. Ask the intended reader which decision the map supports; change one legend, filter or annotation, then recheck interpretation against the same IDs and values.
