## Browser GIS pathway

**Geographic question:** Can capability become discovery without invented mappings?

**Data readiness (B/O + C):** swath/AOI lesson, geometry presets, research-to-STAC lesson, and bounded discovery companion.

**Start from the chapter evidence:** [swath/AOI lesson](08_swaths_orbits_study_area.ipynb) · [geometry presets](https://github.com/jltobias/JupyterLite-Learning-Atlas/blob/main/content/courses/eo-satellite-guide/data/geometry-presets.json) (`courses/eo-satellite-guide/data/geometry-presets.json`) · [research-to-STAC lesson](07_research_to_stac.ipynb) · [bounded discovery companion](../../projects/eo-satellite-guide/index.ipynb)

- **[RustGIS Web](https://web.rustgis.app/):** Use existing declared AOI/preset or learner-attributed AOI; retain property/processing/date interval and actual route states. In RustGIS, import the reviewed geographic export through Add Data; inspect the selected feature IDs and attribute table before styling.
- **[GeoLibre Web](https://web.geolibre.app/):** Use existing declared AOI/preset or learner-attributed AOI; retain property/processing/date interval and actual route states. In GeoLibre, import that same reviewed file, compare a second legend/filter, and reconcile its count and one known attribute with the notebook.

**Handoff:** Use the [GIS workbench](../../gis-workbench.ipynb) to validate/export an attributed WGS84 layer. Download the GeoJSON, method record and review prompt from `exports/`, then import through the chosen app’s file picker. Compare `_atlas_source_id` (or source `GEOID`) with the original IDs; app-generated internal IDs may differ. Return changed results by explicit upload. Apps and the notebook share no files or kernel state. Practice-only Census data remain Fresno 2020, never this subject’s observations.

**Optional AI prompt (approved subject evidence only):** “Reject TOA/SR mismatch and identify unmapped collection IDs. Cite supplied feature IDs/fields, separate observations from assumptions, abstain when evidence is missing, and propose one falsifiable check.” For the common practice package, use its generated schema/summary prompt instead. A human-written answer completes the same task.

**GeoAI opportunity:** Image interpretation needs acquired pixels; capability curation is metadata. Any later fitted exercise must name target/features and spatial/temporal support, reserve a spatial/temporal holdout, inspect leakage and uncertainty, and state evidence that would disprove the interpretation. Styling, thresholds, schema checks and chat alone are not model fitting.

**Human evidence check:** No invented collection/API route/acquisition claim; EarthCODE integration remains proposal. Ask the intended reader which decision the map supports; change one legend, filter or annotation, then recheck interpretation against the same IDs and values.
