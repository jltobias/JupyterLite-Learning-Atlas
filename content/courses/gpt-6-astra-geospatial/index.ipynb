## Browser GIS pathway

**Geographic question:** Can an explanation meet a spatial evidence contract?

**Data readiness (B):** spatial contracts, spatial evaluation, and building geometry.

**Start from the chapter evidence:** [spatial contracts](01_spatial_contracts.ipynb) · [spatial evaluation](09_astra_spatial_eval.ipynb) · [building geometry](https://github.com/jltobias/JupyterLite-Learning-Atlas/blob/main/content/courses/gpt-6-astra-geospatial/data/gaborone_buildings.geojson) (`courses/gpt-6-astra-geospatial/data/gaborone_buildings.geojson`)

- **[RustGIS Web](https://web.rustgis.app/):** Select a few bundled building IDs; retain source/version/time and inspect coverage of required fields. In RustGIS, import the reviewed geographic export through Add Data; inspect the selected feature IDs and attribute table before styling.
- **[GeoLibre Web](https://web.geolibre.app/):** Select a few bundled building IDs; retain source/version/time and inspect coverage of required fields. In GeoLibre, import that same reviewed file, compare a second legend/filter, and reconcile its count and one known attribute with the notebook.

**Handoff:** Use the [GIS workbench](../../gis-workbench.ipynb) to validate/export an attributed WGS84 layer. Download the GeoJSON, method record and review prompt from `exports/`, then import through the chosen app’s file picker. Compare `_atlas_source_id` (or source `GEOID`) with the original IDs; app-generated internal IDs may differ. Return changed results by explicit upload. Apps and the notebook share no files or kernel state. Practice-only Census data remain Fresno 2020, never this subject’s observations.

**Optional AI prompt (approved subject evidence only):** “Answer from supplied fields; abstain on absent occupancy/population. Cite supplied feature IDs/fields, separate observations from assumptions, abstain when evidence is missing, and propose one falsifiable check.” For the common practice package, use its generated schema/summary prompt instead. A human-written answer completes the same task.

**GeoAI opportunity:** Evaluate candidate answers; no unrun benchmark or world-model result. Any later fitted exercise must name target/features and spatial/temporal support, reserve a spatial/temporal holdout, inspect leakage and uncertainty, and state evidence that would disprove the interpretation. Styling, thresholds, schema checks and chat alone are not model fitting.

**Human evidence check:** Schema coverage is not accuracy; audit alignment with separate illustrative TB layers. Ask the intended reader which decision the map supports; change one legend, filter or annotation, then recheck interpretation against the same IDs and values.
