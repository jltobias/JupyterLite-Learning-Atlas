## Browser GIS pathway

**Geographic question:** What must survive a changed-result handoff?

**Data readiness (B/O):** existing reproducible handoff, teaching layer, WorldPop footprints, provenance, and original 2D export companion.

**Start from the chapter evidence:** [existing reproducible handoff](03_rustgis_bridge.ipynb) · [teaching layer](https://github.com/jltobias/JupyterLite-Learning-Atlas/blob/main/content/courses/rustgis/data/demo_features.geojson) (`courses/rustgis/data/demo_features.geojson`) · [WorldPop footprints](https://github.com/jltobias/JupyterLite-Learning-Atlas/blob/main/content/courses/rustgis/data/worldpop_uga_items.geojson) (`courses/rustgis/data/worldpop_uga_items.geojson`) · [provenance](https://github.com/jltobias/JupyterLite-Learning-Atlas/blob/main/content/courses/rustgis/data/worldpop_provenance.json) (`courses/rustgis/data/worldpop_provenance.json`) · [original 2D export companion](../../projects/rustgis/index.ipynb)

- **[RustGIS Web](https://web.rustgis.app/):** Use bundled demo geometry or companion's existing 2D points; export data/method card and manually import into both apps. In RustGIS, import the reviewed geographic export through Add Data; inspect the selected feature IDs and attribute table before styling.
- **[GeoLibre Web](https://web.geolibre.app/):** Use bundled demo geometry or companion's existing 2D points; export data/method card and manually import into both apps. In GeoLibre, import that same reviewed file, compare a second legend/filter, and reconcile its count and one known attribute with the notebook.

**Handoff:** Use the [GIS workbench](../../gis-workbench.ipynb) to validate/export an attributed WGS84 layer. Download the GeoJSON, method record and review prompt from `exports/`, then import through the chosen app’s file picker. Compare `_atlas_source_id` (or source `GEOID`) with the original IDs; app-generated internal IDs may differ. Return changed results by explicit upload. Apps and the notebook share no files or kernel state. Practice-only Census data remain Fresno 2020, never this subject’s observations.

**Optional AI prompt (approved subject evidence only):** “Audit ID/count/bbox/units/source/missingness before and after transfer. Cite supplied feature IDs/fields, separate observations from assumptions, abstain when evidence is missing, and propose one falsifiable check.” For the common practice package, use its generated schema/summary prompt instead. A human-written answer completes the same task.

**GeoAI opportunity:** Reuse existing DBSCAN/holdout as optional GeoAI; projection is not training. Any later fitted exercise must name target/features and spatial/temporal support, reserve a spatial/temporal holdout, inspect leakage and uncertainty, and state evidence that would disprove the interpretation. Styling, thresholds, schema checks and chat alone are not model fitting.

**Human evidence check:** Preserve source pins/properties; no native CPython wheel installed in Pyodide. Ask the intended reader which decision the map supports; change one legend, filter or annotation, then recheck interpretation against the same IDs and values.
