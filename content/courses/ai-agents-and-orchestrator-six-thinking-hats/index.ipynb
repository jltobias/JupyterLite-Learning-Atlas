## Browser GIS pathway

**Geographic question:** Can reviewers separate evidence from values?

**Data readiness (N + optional B):** foundation lesson, orchestrator lab, and original role-coverage exercise.

**Start from the chapter evidence:** [foundation lesson](00_six_thinking_hats_foundations.ipynb) · [orchestrator lab](02_orchestrator_lab.ipynb) · [original role-coverage exercise](../../projects/ai-agents-and-orchestrator-six-thinking-hats/index.ipynb)

- **[RustGIS Web](https://web.rustgis.app/):** Review an existing teaching/practice layer. Keep remarks in a decision log keyed to actual feature/layer IDs. In RustGIS, import the reviewed geographic export through Add Data; inspect the selected feature IDs and attribute table before styling.
- **[GeoLibre Web](https://web.geolibre.app/):** Review an existing teaching/practice layer. Keep remarks in a decision log keyed to actual feature/layer IDs. In GeoLibre, import that same reviewed file, compare a second legend/filter, and reconcile its count and one known attribute with the notebook.

**Handoff:** Use the [GIS workbench](../../gis-workbench.ipynb) to validate/export an attributed WGS84 layer. Download the GeoJSON, method record and review prompt from `exports/`, then import through the chosen app’s file picker. Compare `_atlas_source_id` (or source `GEOID`) with the original IDs; app-generated internal IDs may differ. Return changed results by explicit upload. Apps and the notebook share no files or kernel state. Practice-only Census data remain Fresno 2020, never this subject’s observations.

**Optional AI prompt (approved subject evidence only):** “For each hat cite a supplied feature and label evidence, concern or preference. Cite supplied feature IDs/fields, separate observations from assumptions, abstain when evidence is missing, and propose one falsifiable check.” For the common practice package, use its generated schema/summary prompt instead. A human-written answer completes the same task.

**GeoAI opportunity:** Role orchestration is not spatial fitting. Any later fitted exercise must name target/features and spatial/temporal support, reserve a spatial/temporal holdout, inspect leakage and uncertainty, and state evidence that would disprove the interpretation. Styling, thresholds, schema checks and chat alone are not model fitting.

**Human evidence check:** No invented coordinates for statements; role coverage does not establish correctness. Ask the intended reader which decision the map supports; change one legend, filter or annotation, then recheck interpretation against the same IDs and values.
