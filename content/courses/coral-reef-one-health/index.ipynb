**GeoAI opportunity:** Habitat classification needs actual labelled imagery/validation. Styling, thresholds, schema checks and chat alone are not model fitting.

**Human evidence check:** No invented reef coordinates/name geocoding; missing score remains null. Ask the intended reader which decision the map supports; change one legend, filter or annotation, then recheck interpretation against the same IDs and values.
