## Try it: visualise network degree

The grid below is a tiny invented graph. The labels show the number of direct street connections (degree), not measured accessibility or travel demand.

````python
rows, columns = 3, 4
grid_edges = []
for row in range(rows):
    for column in range(columns):
        node = (row, column)
        if column + 1 < columns:
            grid_edges.append((node, (row, column + 1)))
        if row + 1 < rows:
            grid_edges.append((node, (row + 1, column)))

degree = {}
for left, right in grid_edges:
    degree[left] = degree.get(left, 0) + 1
    degree[right] = degree.get(right, 0) + 1
for row in range(rows):
    print("  ".join(str(degree[(row, column)]) for column in range(columns)))
print("Each digit is degree; each position is one synthetic intersection.")
````

## Native Urbanity visualisations

The official examples include static street-network plots, centrality maps, building and node-attribute views, and an interactive 3D graph view. These require the native package and its geospatial/visualisation stack. Check the current [visualisation API](https://urbanity.readthedocs.io/en/latest/api/visualisation/) for signatures and argument names.

```python
from urbanity.visualisation import (
    plot_street_network,
    plot_network_centrality,
    plot_urban_graph_overview,
    plot_graph,
)

fig = plot_street_network(nodes, edges, title="Study area streets")
fig = plot_network_centrality(nodes, edges, centrality_col="betweenness")
fig = plot_urban_graph_overview(urban_graph.geo_store)
plot_graph(urban_graph.geo_store, urban_graph.edge_store, node_type="building")
```

An effective map should state the study-area boundary, units, data vintage, and whether values are measured, modelled, or illustrative. Avoid implying that 3D height or colour encodes an attribute when it does not.
