# Street-view and satellite imagery

Imagery can add a visual layer of evidence to urban analysis: street-view imagery can support streetscape indicators, while satellite and remote-sensing layers can describe land cover or other surface patterns. Images, derived labels, and provider coverage are not neutral or complete observations. This browser chapter uses a tiny made-up pixel-label array and downloads no imagery.

[Launch this notebook in JupyterLite](https://jltobias.github.io/JupyterLite-Urbanity/lab/lab/index.html?path=notebooks/imagery.ipynb) · [Urbanity satellite API](https://urbanity.readthedocs.io/en/latest/api/satellite/) · [Street-view paper](https://doi.org/10.1177/23998083221138832)

## Try it: count labels in a synthetic image patch

Treat each item as a simplified pixel class. The classes are invented for this exercise. This is not a segmentation model or a result from Mapillary, Mapbox, or Google Earth Engine.

```python
toy_patch = [
    ["tree", "tree", "building", "sky"],
    ["tree", "road", "road", "sky"],
    ["person", "road", "building", "building"],
]

counts = {}
for row in toy_patch:
    for label in row:
        counts[label] = counts.get(label, 0) + 1
pixel_total = sum(counts.values())
for label, count in sorted(counts.items()):
    print(f"{label:10} {count:2} cells ({count / pixel_total:.0%})")
```

A pixel share is not automatically a street-level indicator: camera angle, season, lighting, occlusion, image selection, and model errors all matter. Imagery may also expose people or sensitive places. Use aggregation and governance appropriate to the question, and check provider restrictions before processing or publishing.
