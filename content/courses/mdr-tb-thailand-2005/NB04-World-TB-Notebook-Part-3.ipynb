## STEP 4. Visualizations using Plotly Express

```python
px.scatter(merged_df8_2016, x="GDP-USD", \
           log_x=True, log_y=True, y="e_mort_100k", \
           hover_name="Country", color="IncomeGroup")
```

```python
px.scatter(merged_df8_2016, x="GDP-USD", \
           log_x=True, log_y=True, \
           y="e_mort_tbhiv_100k", \
           hover_name="Country", color="IncomeGroup")
```

```python
px.scatter(merged_df8_2016, x="GDP-USD", y="e_mort_100k", \
           color="IncomeGroup", size="e_pop_num", size_max=60, \
           hover_name="Country")
```

```python
px.scatter_matrix(merged_df8_2016, \
        dimensions=["GDP-USD", "Primary-Completion-Rate", \
                    "e_pop_num", "e_inc_100k", "e_inc_tbhiv_100k"], \
                  hover_name="Country", color="IncomeGroup")
```

```python
px.scatter(merged_df8, x="GDP-USD", y="e_mort_num", animation_frame="Year", \
           animation_group="Country", width=1000,\
           size="e_pop_num", color="Region", hover_name="Country", \
           log_x=True, log_y=True, size_max=80, \
           range_x=[100,120000], range_y=[1,3000000])
```

```python
px.scatter(merged_df8, x="GDP-USD", y="e_inc_100k", animation_frame="Year", \
           animation_group="Country", width=1000,\
           size="e_pop_num", color="Region", hover_name="Country", \
           log_x=True, log_y=True, size_max=80,\
           range_x=[50,120000], range_y=[1,4000])
```

```python
px.choropleth(merged_df8, locations="ISO_Alpha", \
              color="e_mort_100k", hover_name="Country", \
              animation_frame="Year",
              color_continuous_scale=px.colors.sequential.OrRd, \
              range_color=[0,600])
```

```python
px.choropleth(merged_df8, locations="ISO_Alpha", \
              color="e_inc_tbhiv_100k", hover_name="Country", \
              animation_frame="Year",
              color_continuous_scale=px.colors.sequential.OrRd, \
              range_color=[0,1000])
```
