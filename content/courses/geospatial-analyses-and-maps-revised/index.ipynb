## Browser GIS pathway

**Geographic question:** How do counts and within-district shares change comparison?

**Data readiness (B):** Montreal choropleth notebook, 58 district features, election attributes, and dataset notes.

**Start from the chapter evidence:** [Montreal choropleth notebook](notebooks/01_montreal_choropleth_real_boundaries.ipynb) · [58 district features](https://github.com/jltobias/JupyterLite-Learning-Atlas/blob/main/content/courses/geospatial-analyses-and-maps-revised/data/montreal_districts.geojson) (`courses/geospatial-analyses-and-maps-revised/data/montreal_districts.geojson`) · [election attributes](https://github.com/jltobias/JupyterLite-Learning-Atlas/blob/main/content/courses/geospatial-analyses-and-maps-revised/data/election.csv) (`courses/geospatial-analyses-and-maps-revised/data/election.csv`) · [dataset notes](https://github.com/jltobias/JupyterLite-Learning-Atlas/blob/main/content/courses/geospatial-analyses-and-maps-revised/notices/SOURCES.md) (`courses/geospatial-analyses-and-maps-revised/notices/SOURCES.md`)

- **[RustGIS Web](https://web.rustgis.app/):** Join bundled actual Montreal boundaries/election table by documented IDs; compare fixed-legend counts/shares. In RustGIS, import the reviewed geographic export through Add Data; inspect the selected feature IDs and attribute table before styling.
- **[GeoLibre Web](https://web.geolibre.app/):** Join bundled actual Montreal boundaries/election table by documented IDs; compare fixed-legend counts/shares. In GeoLibre, import that same reviewed file, compare a second legend/filter, and reconcile its count and one known attribute with the notebook.

**Handoff:** Use the [GIS workbench](../../gis-workbench.ipynb) to validate/export an attributed WGS84 layer. Download the GeoJSON, method record and review prompt from `exports/`, then import through the chosen app’s file picker. Compare `_atlas_source_id` (or source `GEOID`) with the original IDs; app-generated internal IDs may differ. Return changed results by explicit upload. Apps and the notebook share no files or kernel state. Practice-only Census data remain Fresno 2020, never this subject’s observations.

**Optional AI prompt (approved subject evidence only):** “Explain one denominator/district-size issue using IDs. Cite supplied feature IDs/fields, separate observations from assumptions, abstain when evidence is missing, and propose one falsifiable check.” For the common practice package, use its generated schema/summary prompt instead. A human-written answer completes the same task.

**GeoAI opportunity:** Aggregation is not GeoAI; prediction requires a target and holdout. Any later fitted exercise must name target/features and spatial/temporal support, reserve a spatial/temporal holdout, inspect leakage and uncertainty, and state evidence that would disprove the interpretation. Styling, thresholds, schema checks and chat alone are not model fitting.

**Human evidence check:** Preserve actual boundaries/notices and reconcile matched counts/totals. Ask the intended reader which decision the map supports; change one legend, filter or annotation, then recheck interpretation against the same IDs and values.
