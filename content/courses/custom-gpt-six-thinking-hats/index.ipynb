## Browser GIS pathway

**Geographic question:** Does valid review JSON establish valid claims?

**Data readiness (N + optional B):** JSON capture lesson, global-health example, and schema-check companion.

**Start from the chapter evidence:** [JSON capture lesson](02_prompt_and_capture.ipynb) · [global-health example](https://github.com/jltobias/JupyterLite-Learning-Atlas/blob/main/content/courses/custom-gpt-six-thinking-hats/examples/global_health.json) (`courses/custom-gpt-six-thinking-hats/examples/global_health.json`) · [schema-check companion](../../projects/custom-gpt-six-thinking-hats/index.ipynb)

- **[RustGIS Web](https://web.rustgis.app/):** Paste/manually author JSON about an existing practice layer; retain ID-keyed matched remarks and nonspatial unmatched remarks. In RustGIS, import the reviewed geographic export through Add Data; inspect the selected feature IDs and attribute table before styling.
- **[GeoLibre Web](https://web.geolibre.app/):** Paste/manually author JSON about an existing practice layer; retain ID-keyed matched remarks and nonspatial unmatched remarks. In GeoLibre, import that same reviewed file, compare a second legend/filter, and reconcile its count and one known attribute with the notebook.

**Handoff:** Use the [GIS workbench](../../gis-workbench.ipynb) to validate/export an attributed WGS84 layer. Download the GeoJSON, method record and review prompt from `exports/`, then import through the chosen app’s file picker. Compare `_atlas_source_id` (or source `GEOID`) with the original IDs; app-generated internal IDs may differ. Return changed results by explicit upload. Apps and the notebook share no files or kernel state. Practice-only Census data remain Fresno 2020, never this subject’s observations.

**Optional AI prompt (approved subject evidence only):** “Classify each claim as supported, judgment or unsupported. Cite supplied feature IDs/fields, separate observations from assumptions, abstain when evidence is missing, and propose one falsifiable check.” For the common practice package, use its generated schema/summary prompt instead. A human-written answer completes the same task.

**GeoAI opportunity:** Chat/schema validation is not GeoAI. Any later fitted exercise must name target/features and spatial/temporal support, reserve a spatial/temporal holdout, inspect leakage and uncertainty, and state evidence that would disprove the interpretation. Styling, thresholds, schema checks and chat alone are not model fitting.

**Human evidence check:** Missing and empty hats differ; no invented statement coordinates or assumed plugin access. Ask the intended reader which decision the map supports; change one legend, filter or annotation, then recheck interpretation against the same IDs and values.
