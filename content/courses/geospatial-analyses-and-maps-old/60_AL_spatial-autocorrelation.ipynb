## Learning move

Use `Spatial Autocorrelation` to answer a concrete question: *what changes when location is treated as data rather than decoration?*

```python
from pathlib import Path
import sys
for p in [Path.cwd(), *Path.cwd().parents]:
    if (p / 'spatial_lite.py').exists():
        sys.path.insert(0, str(p))
        break
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import spatial_lite as sl

pd.set_option('display.max_columns', 20)
```

```python
df = sl.load_dataset('landcover_grid.csv')
sl.quick_summary(df)
```

## 1. See the spatial pattern

Before running an algorithm, inspect scale, outliers, and geography. This lowers cognitive load and makes later transformations easier to diagnose.

```python
grid = df.copy()
sl.plot_grid(grid, 'elevation_m', 'spatial-autocorrelation: elevation surface')
sl.plot_grid(grid, 'ndvi', 'spatial-autocorrelation: vegetation index')
summary = grid.groupby('landcover')[['elevation_m','ndvi']].mean().round(3)
display(summary)
```

## 2. Apply the analysis idea

The code below uses small, transparent calculations that run in Pyodide without compiled geospatial dependencies.

```python
sids = sl.load_dataset('sids_nc.csv')
for radius in [60, 100, 160]:
    print(f'Moran I for SIDS rate within {radius} km:', round(sl.morans_i(sids, 'sids_rate', threshold_km=radius), 3))
```
