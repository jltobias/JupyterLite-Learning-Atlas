## Learning move

Use `Directions` to answer a concrete question: *what changes when location is treated as data rather than decoration?*

```python
from pathlib import Path
import sys
for p in [Path.cwd(), *Path.cwd().parents]:
    if (p / 'spatial_lite.py').exists():
        sys.path.insert(0, str(p))
        break
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import spatial_lite as sl

pd.set_option('display.max_columns', 20)
```

```python
df = sl.load_dataset('nyc_taxi_sample.csv')
sl.quick_summary(df)
```

## 1. See the spatial pattern

Before running an algorithm, inspect scale, outliers, and geography. This lowers cognitive load and makes later transformations easier to diagnose.

```python
trips = df.copy()
trips['speed_kmh'] = trips['distance_km'] / (trips['duration_min']/60)
display(trips.groupby(['pickup_zone','dropoff_zone'])[['distance_km','fare_usd','duration_min']].mean().round(2).head(12))
# pickup geography as a map-like scatter
sl.plot_points(trips.rename(columns={'pickup_lon':'lon','pickup_lat':'lat'}), color='fare_usd', title='directions: taxi pickup geography')
```

## 2. Apply the analysis idea

The code below uses small, transparent calculations that run in Pyodide without compiled geospatial dependencies.

```python
trips=sl.load_dataset('nyc_taxi_sample.csv').head(20).copy()
trips['straight_line_km']=sl.haversine(trips.pickup_lon,trips.pickup_lat,trips.dropoff_lon,trips.dropoff_lat)
display(trips[['pickup_zone','dropoff_zone','straight_line_km','duration_min','fare_usd']].round(2).head(10))
```
