## Browser GIS pathway

**Geographic question:** Are rates compatible in measure, year and denominator?

**Data readiness (B):** burden pipeline template, synthetic country-point layer, country reference, and ethics lesson.

**Start from the chapter evidence:** [burden pipeline template](01_gbd_data_pipeline_template.ipynb) · [synthetic country-point layer](https://github.com/jltobias/JupyterLite-Learning-Atlas/blob/main/content/courses/global-health-gis/data/gbd_country_points_2021.geojson) (`courses/global-health-gis/data/gbd_country_points_2021.geojson`) · [country reference](https://github.com/jltobias/JupyterLite-Learning-Atlas/blob/main/content/courses/global-health-gis/data/country_reference.csv) (`courses/global-health-gis/data/country_reference.csv`) · [ethics lesson](05_public_data_sources_and_ethics.ipynb)

- **[RustGIS Web](https://web.rustgis.app/):** Select one explicit slice of bundled synthetic country points; inspect ISO3, source, age, sex and intervals. In RustGIS, import the reviewed geographic export through Add Data; inspect the selected feature IDs and attribute table before styling.
- **[GeoLibre Web](https://web.geolibre.app/):** Select one explicit slice of bundled synthetic country points; inspect ISO3, source, age, sex and intervals. In GeoLibre, import that same reviewed file, compare a second legend/filter, and reconcile its count and one known attribute with the notebook.

**Handoff:** Use the [GIS workbench](../../gis-workbench.ipynb) to validate/export an attributed WGS84 layer. Download the GeoJSON, method record and review prompt from `exports/`, then import through the chosen app’s file picker. Compare `_atlas_source_id` (or source `GEOID`) with the original IDs; app-generated internal IDs may differ. Return changed results by explicit upload. Apps and the notebook share no files or kernel state. Practice-only Census data remain Fresno 2020, never this subject’s observations.

**Optional AI prompt (approved subject evidence only):** “List incompatible measures before summarizing. Cite supplied feature IDs/fields, separate observations from assumptions, abstain when evidence is missing, and propose one falsifiable check.” For the common practice package, use its generated schema/summary prompt instead. A human-written answer completes the same task.

**GeoAI opportunity:** Observed labels/validation needed for spatial inference. Any later fitted exercise must name target/features and spatial/temporal support, reserve a spatial/temporal holdout, inspect leakage and uncertainty, and state evidence that would disprove the interpretation. Styling, thresholds, schema checks and chat alone are not model fitting.

**Human evidence check:** Retain SYNTHETIC_DEMO_NOT_IHME; do not casually population-weight age-standardized rates. Ask the intended reader which decision the map supports; change one legend, filter or annotation, then recheck interpretation against the same IDs and values.
