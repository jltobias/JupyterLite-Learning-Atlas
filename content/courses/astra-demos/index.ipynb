## Browser GIS pathway

**Geographic question:** Which generated-scene claims transfer to real geography?

**Data readiness (O/B):** terrain/water source, interpolation lesson, and procedural-terrain companion.

**Start from the chapter evidence:** [terrain/water source](07_terrain_water.ipynb) · [interpolation lesson](14_spatial_interpolation.ipynb) · [procedural-terrain companion](../../projects/astra-demos/index.ipynb)

- **[RustGIS Web](https://web.rustgis.app/):** Keep procedural terrain generated/nonspatial. Practice only with learner-attributed terrain-summary data or shared tracts. In RustGIS, import the reviewed geographic export through Add Data; inspect the selected feature IDs and attribute table before styling.
- **[GeoLibre Web](https://web.geolibre.app/):** Keep procedural terrain generated/nonspatial. Practice only with learner-attributed terrain-summary data or shared tracts. In GeoLibre, import that same reviewed file, compare a second legend/filter, and reconcile its count and one known attribute with the notebook.

**Handoff:** Use the [GIS workbench](../../gis-workbench.ipynb) to validate/export an attributed WGS84 layer. Download the GeoJSON, method record and review prompt from `exports/`, then import through the chosen app’s file picker. Compare `_atlas_source_id` (or source `GEOID`) with the original IDs; app-generated internal IDs may differ. Return changed results by explicit upload. Apps and the notebook share no files or kernel state. Practice-only Census data remain Fresno 2020, never this subject’s observations.

**Optional AI prompt (approved subject evidence only):** “Distinguish generated, observed and unknown attributes; explain smoothing loss. Cite supplied feature IDs/fields, separate observations from assumptions, abstain when evidence is missing, and propose one falsifiable check.” For the common practice package, use its generated schema/summary prompt instead. A human-written answer completes the same task.

**GeoAI opportunity:** Interpolation/model comparison needs support and held-out evidence. Any later fitted exercise must name target/features and spatial/temporal support, reserve a spatial/temporal holdout, inspect leakage and uncertainty, and state evidence that would disprove the interpretation. Styling, thresholds, schema checks and chat alone are not model fitting.

**Human evidence check:** No geographic coordinates assigned to the generated world; no benchmark result claimed. Ask the intended reader which decision the map supports; change one legend, filter or annotation, then recheck interpretation against the same IDs and values.
