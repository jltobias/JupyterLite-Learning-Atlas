## Three views, three questions

A 2D map is useful for location and comparison. A 3D terrain map reveals relief. A scene can combine surfaces and objects. A globe avoids the rectangular map's polar stretching but hides the far side. None is universally best.

```python
map_outlines(title="2D longitude–latitude reference view")
plt.show()
interactive(surface(exaggeration=1))
```

## Project onto a unit sphere

We convert longitude λ and latitude φ to x=cosφ cosλ, y=cosφ sinλ, z=sinφ. The unit sphere is a visualization, not an ellipsoidal Earth measurement model. Country outlines come from Natural Earth; the globe itself is generated mathematically.

```python
longitude, latitude = np.deg2rad([-105, 40])
point = np.array([np.cos(latitude)*np.cos(longitude), np.cos(latitude)*np.sin(longitude), np.sin(latitude)])
assert np.isclose(np.linalg.norm(point), 1)
print("Unit sphere location:", point)
interactive(globe())
```

## Add an explicit scene object

A route above a surface is not automatically visible, traversable, or safe. We add an invented survey transect only to practice layering. The height offset is a visual aid disclosed in the legend.

```python
import plotly.graph_objects as go
z = terrain(); columns = np.arange(96); row = 48
xs = TRANSFORM[0] + (columns+.5)*30
ys = np.full(96, TRANSFORM[3]-(row+.5)*30)
fig = surface(z, exaggeration=2)
fig.add_trace(go.Scatter3d(x=xs, y=ys, z=z[row]+15, mode="lines", line=dict(color="orange", width=7),
                           name="Invented transect: displayed 15 m above terrain"))
interactive(fig)
```

## Provide a non-3D reading path

WebGL can be unavailable on some devices. A labelled profile preserves the transect information without requiring 3D interaction. Similar alternatives should accompany every scene used in an assessment.

```python
fig, ax = plt.subplots(figsize=(8, 3))
ax.plot((xs-xs[0])/1000, z[row])
ax.set(xlabel="Distance along transect (km)", ylabel="Elevation (m)", title="Synthetic transect profile; no display offset")
plt.show()
```
