## Browser GIS pathway

**Geographic question:** How can unverified signals stay visible without appearing confirmed?

**Data readiness (O + C):** common-picture source, verification lesson, sitrep lesson, and original report-state exercise.

**Start from the chapter evidence:** [common-picture source](02_2d_common_picture.ipynb) · [verification lesson](06_cameras_and_verification.ipynb) · [sitrep lesson](08_priorities_and_sitrep.ipynb) · [original report-state exercise](../../projects/osiris/index.ipynb)

- **[RustGIS Web](https://web.rustgis.app/):** Use evidence-state fields of a learner-attributed geographic export if supplied; otherwise practice only and retain original report list nonspatial. In RustGIS, import the reviewed geographic export through Add Data; inspect the selected feature IDs and attribute table before styling.
- **[GeoLibre Web](https://web.geolibre.app/):** Use evidence-state fields of a learner-attributed geographic export if supplied; otherwise practice only and retain original report list nonspatial. In GeoLibre, import that same reviewed file, compare a second legend/filter, and reconcile its count and one known attribute with the notebook.

**Handoff:** Use the [GIS workbench](../../gis-workbench.ipynb) to validate/export an attributed WGS84 layer. Download the GeoJSON, method record and review prompt from `exports/`, then import through the chosen app’s file picker. Compare `_atlas_source_id` (or source `GEOID`) with the original IDs; app-generated internal IDs may differ. Return changed results by explicit upload. Apps and the notebook share no files or kernel state. Practice-only Census data remain Fresno 2020, never this subject’s observations.

**Optional AI prompt (approved subject evidence only):** “Draft evidence-state-labelled briefing without promoting plausibility to fact. Cite supplied feature IDs/fields, separate observations from assumptions, abstain when evidence is missing, and propose one falsifiable check.” For the common practice package, use its generated schema/summary prompt instead. A human-written answer completes the same task.

**GeoAI opportunity:** Anomaly/vision models need verified inputs and separate evaluation. Any later fitted exercise must name target/features and spatial/temporal support, reserve a spatial/temporal holdout, inspect leakage and uncertainty, and state evidence that would disprove the interpretation. Styling, thresholds, schema checks and chat alone are not model fitting.

**Human evidence check:** No newly invented camera, live signal or operational report geometry. Ask the intended reader which decision the map supports; change one legend, filter or annotation, then recheck interpretation against the same IDs and values.
