## Browser GIS pathway

**Geographic question:** Whose preferences change the next experiment?

**Data readiness (O/N + C):** digital-twin source, privacy lesson, and priority-weight companion.

**Start from the chapter evidence:** [digital-twin source](08_digital_twin_global_health.ipynb) · [privacy lesson](09_geoprivacy_map_encryption.ipynb) · [priority-weight companion](../../projects/rewiring-global-health/index.ipynb)

- **[RustGIS Web](https://web.rustgis.app/):** Review existing practice layer with a weight/decision log. Keep privacy review/workshop as actions, not geographic sites. In RustGIS, import the reviewed geographic export through Add Data; inspect the selected feature IDs and attribute table before styling.
- **[GeoLibre Web](https://web.geolibre.app/):** Review existing practice layer with a weight/decision log. Keep privacy review/workshop as actions, not geographic sites. In GeoLibre, import that same reviewed file, compare a second legend/filter, and reconcile its count and one known attribute with the notebook.

**Handoff:** Use the [GIS workbench](../../gis-workbench.ipynb) to validate/export an attributed WGS84 layer. Download the GeoJSON, method record and review prompt from `exports/`, then import through the chosen app’s file picker. Compare `_atlas_source_id` (or source `GEOID`) with the original IDs; app-generated internal IDs may differ. Return changed results by explicit upload. Apps and the notebook share no files or kernel state. Practice-only Census data remain Fresno 2020, never this subject’s observations.

**Optional AI prompt (approved subject evidence only):** “Separate changed preferences, observations and community evidence. Cite supplied feature IDs/fields, separate observations from assumptions, abstain when evidence is missing, and propose one falsifiable check.” For the common practice package, use its generated schema/summary prompt instead. A human-written answer completes the same task.

**GeoAI opportunity:** Uncalibrated world models remain continuations; ranking is not prediction. Any later fitted exercise must name target/features and spatial/temporal support, reserve a spatial/temporal holdout, inspect leakage and uncertainty, and state evidence that would disprove the interpretation. Styling, thresholds, schema checks and chat alone are not model fitting.

**Human evidence check:** No fake action locations; encryption exercise is not managed key service. Ask the intended reader which decision the map supports; change one legend, filter or annotation, then recheck interpretation against the same IDs and values.
