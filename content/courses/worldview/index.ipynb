## Browser GIS pathway

**Geographic question:** How can equal allocation leave unmet demand?

**Data readiness (B):** response source, synthetic districts, facilities, network, and allocation companion.

**Start from the chapter evidence:** [response source](04_eoc_response.ipynb) · [synthetic districts](https://github.com/jltobias/JupyterLite-Learning-Atlas/blob/main/content/courses/worldview/data/districts.geojson) (`courses/worldview/data/districts.geojson`) · [facilities](https://github.com/jltobias/JupyterLite-Learning-Atlas/blob/main/content/courses/worldview/data/facilities.geojson) (`courses/worldview/data/facilities.geojson`) · [network](https://github.com/jltobias/JupyterLite-Learning-Atlas/blob/main/content/courses/worldview/data/network.json) (`courses/worldview/data/network.json`) · [allocation companion](../../projects/worldview/index.ipynb)

- **[RustGIS Web](https://web.rustgis.app/):** Inspect existing synthetic district/facility/network fields and bounded scenarios. Keep A/B/C companion values separate without crosswalk. In RustGIS, import the reviewed geographic export through Add Data; inspect the selected feature IDs and attribute table before styling.
- **[GeoLibre Web](https://web.geolibre.app/):** Inspect existing synthetic district/facility/network fields and bounded scenarios. Keep A/B/C companion values separate without crosswalk. In GeoLibre, import that same reviewed file, compare a second legend/filter, and reconcile its count and one known attribute with the notebook.

**Handoff:** Use the [GIS workbench](../../gis-workbench.ipynb) to validate/export an attributed WGS84 layer. Download the GeoJSON, method record and review prompt from `exports/`, then import through the chosen app’s file picker. Compare `_atlas_source_id` (or source `GEOID`) with the original IDs; app-generated internal IDs may differ. Return changed results by explicit upload. Apps and the notebook share no files or kernel state. Practice-only Census data remain Fresno 2020, never this subject’s observations.

**Optional AI prompt (approved subject evidence only):** “Audit fairness, stock conservation and missing logistics. Cite supplied feature IDs/fields, separate observations from assumptions, abstain when evidence is missing, and propose one falsifiable check.” For the common practice package, use its generated schema/summary prompt instead. A human-written answer completes the same task.

**GeoAI opportunity:** Network optimization/inference are separate continuations. Any later fitted exercise must name target/features and spatial/temporal support, reserve a spatial/temporal holdout, inspect leakage and uncertainty, and state evidence that would disprove the interpretation. Styling, thresholds, schema checks and chat alone are not model fitting.

**Human evidence check:** No unsupported join or real emergency claim; allocation remains within stock. Ask the intended reader which decision the map supports; change one legend, filter or annotation, then recheck interpretation against the same IDs and values.
