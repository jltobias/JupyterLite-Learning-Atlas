## Browser GIS pathway

**Geographic question:** How do coverage, concept and partial-cell support affect totals?

**Data readiness (B):** real population lesson, coverage lesson, real export lesson, native-count JSON, and independent oracle.

**Start from the chapter evidence:** [real population lesson](01_real_population.ipynb) · [coverage lesson](02_real_coverage.ipynb) · [real export lesson](03_real_exports.ipynb) · [native-count JSON](https://github.com/jltobias/JupyterLite-Learning-Atlas/blob/main/content/courses/population-data-comparison/population.json) (`courses/population-data-comparison/population.json`) · [independent oracle](https://github.com/jltobias/JupyterLite-Learning-Atlas/blob/main/content/courses/population-data-comparison/independent_oracle.json) (`courses/population-data-comparison/independent_oracle.json`)

- **[RustGIS Web](https://web.rustgis.app/):** Reuse existing stdlib helper, population.json and independent oracle. Export one actual native GHS-POP count rectangle as lesson 03 does. In RustGIS, import the reviewed geographic export through Add Data; inspect the selected feature IDs and attribute table before styling.
- **[GeoLibre Web](https://web.geolibre.app/):** Reuse existing stdlib helper, population.json and independent oracle. Export one actual native GHS-POP count rectangle as lesson 03 does. In GeoLibre, import that same reviewed file, compare a second legend/filter, and reconcile its count and one known attribute with the notebook.

**Handoff:** Use the [GIS workbench](../../gis-workbench.ipynb) to validate/export an attributed WGS84 layer. Download the GeoJSON, method record and review prompt from `exports/`, then import through the chosen app’s file picker. Compare `_atlas_source_id` (or source `GEOID`) with the original IDs; app-generated internal IDs may differ. Return changed results by explicit upload. Apps and the notebook share no files or kernel state. Practice-only Census data remain Fresno 2020, never this subject’s observations.

**Optional AI prompt (approved subject evidence only):** “Separate coverage, date, ambient/residential concept and allocation assumptions. Cite supplied feature IDs/fields, separate observations from assumptions, abstain when evidence is missing, and propose one falsifiable check.” For the common practice package, use its generated schema/summary prompt instead. A human-written answer completes the same task.

**GeoAI opportunity:** Population model comparison is meaningful; exporting one cell is not training. Any later fitted exercise must name target/features and spatial/temporal support, reserve a spatial/temporal holdout, inspect leakage and uncertainty, and state evidence that would disprove the interpretation. Styling, thresholds, schema checks and chat alone are not model fitting.

**Human evidence check:** Preserve all metadata/CC BY notices; NoData differs from zero; oracle totals reproduce; no giant export. Ask the intended reader which decision the map supports; change one legend, filter or annotation, then recheck interpretation against the same IDs and values.
