**Optional AI prompt (approved subject evidence only):** “Find multi-country double counting and incompatible denominators. Cite supplied feature IDs/fields, separate observations from assumptions, abstain when evidence is missing, and propose one falsifiable check.” For the common practice package, use its generated schema/summary prompt instead. **GeoAI opportunity:** Patient/mutation modelling requires approved inputs; no clinical inference. Any later fitted exercise must name target/features and spatial/temporal support, reserve a spatial/temporal holdout, inspect leakage and uncertainty, and state evidence that would disprove the interpretation. Styling, thresholds, schema checks and chat alone are not model fitting.
