## Browser GIS pathway

**Geographic question:** What do ties and outside-domain records preserve?

**Data readiness (— O + B/C continuation):** finite nearest-site companion, source Voronoi construction, and source building overlay.

**Start from the chapter evidence:** [finite nearest-site companion](../../projects/cholera-1854-advanced-2/index.ipynb) · [source Voronoi construction](NB03-Constructing-Voronoi-Polygons.ipynb) · [source building overlay](NB05-Buildings-in-Voronoi-Cells.ipynb)

- **[RustGIS Web](https://web.rustgis.app/):** Keep finite planar assignments. Geographic polygons/overlays require supported native source or validated attributed export; practice only otherwise. In RustGIS, import the reviewed geographic export through Add Data; inspect the selected feature IDs and attribute table before styling.
- **[GeoLibre Web](https://web.geolibre.app/):** Keep finite planar assignments. Geographic polygons/overlays require supported native source or validated attributed export; practice only otherwise. In GeoLibre, import that same reviewed file, compare a second legend/filter, and reconcile its count and one known attribute with the notebook.

**Handoff:** Use the [GIS workbench](../../gis-workbench.ipynb) to validate/export an attributed WGS84 layer. Download the GeoJSON, method record and review prompt from `exports/`, then import through the chosen app’s file picker. Compare `_atlas_source_id` (or source `GEOID`) with the original IDs; app-generated internal IDs may differ. Return changed results by explicit upload. Apps and the notebook share no files or kernel state. Practice-only Census data remain Fresno 2020, never this subject’s observations.

**Optional AI prompt (approved subject evidence only):** “Separate sample counts, area, buildings and population coverage. Cite supplied feature IDs/fields, separate observations from assumptions, abstain when evidence is missing, and propose one falsifiable check.” For the common practice package, use its generated schema/summary prompt instead. A human-written answer completes the same task.

**GeoAI opportunity:** Exact nearest-site construction is deterministic, not population inference. Any later fitted exercise must name target/features and spatial/temporal support, reserve a spatial/temporal holdout, inspect leakage and uncertainty, and state evidence that would disprove the interpretation. Styling, thresholds, schema checks and chat alone are not model fitting.

**Human evidence check:** Tie stays tie, outside stays outside; no fake geographic Voronoi output. Ask the intended reader which decision the map supports; change one legend, filter or annotation, then recheck interpretation against the same IDs and values.
