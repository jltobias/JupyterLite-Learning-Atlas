## Browser GIS pathway

**Geographic question:** When does a wider gate trade unmatched for ambiguity?

**Data readiness (B/O + C):** observed-ocean lesson, 1-degree historical activity, derived manifest, and matching-ambiguity companion.

**Start from the chapter evidence:** [observed-ocean lesson](01_observed_ocean.ipynb) · [1-degree historical activity](https://github.com/jltobias/JupyterLite-Learning-Atlas/blob/main/content/courses/global-fishing-watch-and-dark-vessels/data/sar_activity_1deg.csv) (`courses/global-fishing-watch-and-dark-vessels/data/sar_activity_1deg.csv`) · [derived manifest](https://github.com/jltobias/JupyterLite-Learning-Atlas/blob/main/content/courses/global-fishing-watch-and-dark-vessels/data/derived_manifest.json) (`courses/global-fishing-watch-and-dark-vessels/data/derived_manifest.json`) · [matching-ambiguity companion](../../projects/global-fishing-watch-and-dark-vessels/index.ipynb)

- **[RustGIS Web](https://web.rustgis.app/):** Export fixed small window of bundled historical 1-degree activity as documented centers/rectangles; keep original planar gates separate. In RustGIS, import the reviewed geographic export through Add Data; inspect the selected feature IDs and attribute table before styling.
- **[GeoLibre Web](https://web.geolibre.app/):** Export fixed small window of bundled historical 1-degree activity as documented centers/rectangles; keep original planar gates separate. In GeoLibre, import that same reviewed file, compare a second legend/filter, and reconcile its count and one known attribute with the notebook.

**Handoff:** Use the [GIS workbench](../../gis-workbench.ipynb) to validate/export an attributed WGS84 layer. Download the GeoJSON, method record and review prompt from `exports/`, then import through the chosen app’s file picker. Compare `_atlas_source_id` (or source `GEOID`) with the original IDs; app-generated internal IDs may differ. Return changed results by explicit upload. Apps and the notebook share no files or kernel state. Practice-only Census data remain Fresno 2020, never this subject’s observations.

**Optional AI prompt (approved subject evidence only):** “Explain why unmatched detections imply neither wrongdoing nor catch. Cite supplied feature IDs/fields, separate observations from assumptions, abstain when evidence is missing, and propose one falsifiable check.” For the common practice package, use its generated schema/summary prompt instead. A human-written answer completes the same task.

**GeoAI opportunity:** SAR detection/matching models are source continuations; gate rule is deterministic. Any later fitted exercise must name target/features and spatial/temporal support, reserve a spatial/temporal holdout, inspect leakage and uncertainty, and state evidence that would disprove the interpretation. Styling, thresholds, schema checks and chat alone are not model fitting.

**Human evidence check:** Retain 2017-2021 normalized units and deposit/platform rights distinction; no vessel tracks invented. Ask the intended reader which decision the map supports; change one legend, filter or annotation, then recheck interpretation against the same IDs and values.
