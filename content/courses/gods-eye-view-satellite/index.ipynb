## Browser GIS pathway

**Geographic question:** Does possible coverage establish usable observation?

**Data readiness (B/O):** sensor footprint lesson, Riverbend synthetic sectors, and uncertainty lesson.

**Start from the chapter evidence:** [sensor footprint lesson](02_sensor_footprint.ipynb) · [Riverbend synthetic sectors](https://github.com/jltobias/JupyterLite-Learning-Atlas/blob/main/content/courses/gods-eye-view-satellite/data/riverbend.geojson) (`courses/gods-eye-view-satellite/data/riverbend.geojson`) · [uncertainty lesson](11_spatial_uncertainty.ipynb)

- **[RustGIS Web](https://web.rustgis.app/):** Inspect existing synthetic Riverbend sectors and idealized footprint assumptions. Actual footprint requires a documented provider/learner export. In RustGIS, import the reviewed geographic export through Add Data; inspect the selected feature IDs and attribute table before styling.
- **[GeoLibre Web](https://web.geolibre.app/):** Inspect existing synthetic Riverbend sectors and idealized footprint assumptions. Actual footprint requires a documented provider/learner export. In GeoLibre, import that same reviewed file, compare a second legend/filter, and reconcile its count and one known attribute with the notebook.

**Handoff:** Use the [GIS workbench](../../gis-workbench.ipynb) to validate/export an attributed WGS84 layer. Download the GeoJSON, method record and review prompt from `exports/`, then import through the chosen app’s file picker. Compare `_atlas_source_id` (or source `GEOID`) with the original IDs; app-generated internal IDs may differ. Return changed results by explicit upload. Apps and the notebook share no files or kernel state. Practice-only Census data remain Fresno 2020, never this subject’s observations.

**Optional AI prompt (approved subject evidence only):** “List cloud/time/resolution/acquisition evidence absent from a coverage claim. Cite supplied feature IDs/fields, separate observations from assumptions, abstain when evidence is missing, and propose one falsifiable check.” For the common practice package, use its generated schema/summary prompt instead. A human-written answer completes the same task.

**GeoAI opportunity:** Image interpretation needs actual pixels and evaluation. Any later fitted exercise must name target/features and spatial/temporal support, reserve a spatial/temporal holdout, inspect leakage and uncertainty, and state evidence that would disprove the interpretation. Styling, thresholds, schema checks and chat alone are not model fitting.

**Human evidence check:** No invented ephemeris, acquired imagery or real observation claim. Ask the intended reader which decision the map supports; change one legend, filter or annotation, then recheck interpretation against the same IDs and values.
