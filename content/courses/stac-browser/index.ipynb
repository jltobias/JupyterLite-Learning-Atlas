## Browser GIS pathway

**Geographic question:** Is a selected footprint a usable downloaded raster?

**Data readiness (O + B/C continuation):** metadata-selection companion, RustGIS discovery handoff, GeoLibre handoff, and catalog registry.

**Start from the chapter evidence:** [metadata-selection companion](../../projects/stac-browser/index.ipynb) · [RustGIS discovery handoff](04_discover_download_rustgis.ipynb) · [GeoLibre handoff](05_discover_download_geolibre.ipynb) · [catalog registry](https://github.com/jltobias/JupyterLite-Learning-Atlas/blob/main/content/courses/stac-browser/public-catalogs.json) (`courses/stac-browser/public-catalogs.json`)

- **[RustGIS Web](https://web.rustgis.app/):** Reuse existing RustGIS WorldPop item footprints/provenance for metadata-only practice; keep toy companion metadata nonspatial. In RustGIS, import the reviewed geographic export through Add Data; inspect the selected feature IDs and attribute table before styling.
- **[GeoLibre Web](https://web.geolibre.app/):** Reuse existing RustGIS WorldPop item footprints/provenance for metadata-only practice; keep toy companion metadata nonspatial. In GeoLibre, import that same reviewed file, compare a second legend/filter, and reconcile its count and one known attribute with the notebook.

**Handoff:** Use the [GIS workbench](../../gis-workbench.ipynb) to validate/export an attributed WGS84 layer. Download the GeoJSON, method record and review prompt from `exports/`, then import through the chosen app’s file picker. Compare `_atlas_source_id` (or source `GEOID`) with the original IDs; app-generated internal IDs may differ. Return changed results by explicit upload. Apps and the notebook share no files or kernel state. Practice-only Census data remain Fresno 2020, never this subject’s observations.

**Optional AI prompt (approved subject evidence only):** “Separate collection/item/asset/license/unknown cloud cover. Cite supplied feature IDs/fields, separate observations from assumptions, abstain when evidence is missing, and propose one falsifiable check.” For the common practice package, use its generated schema/summary prompt instead. A human-written answer completes the same task.

**GeoAI opportunity:** Optional GeoAI triage needs actual data and validation. Any later fitted exercise must name target/features and spatial/temporal support, reserve a spatial/temporal holdout, inspect leakage and uncertainty, and state evidence that would disprove the interpretation. Styling, thresholds, schema checks and chat alone are not model fitting.

**Human evidence check:** No fabricated footprints, undocumented API/deep links or pixel-download claim. Ask the intended reader which decision the map supports; change one legend, filter or annotation, then recheck interpretation against the same IDs and values.
