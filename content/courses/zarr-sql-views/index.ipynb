## Browser GIS pathway

**Geographic question:** Which threshold changes retained spatial evidence?

**Data readiness (B/O + C):** AR6 grid lesson, grid summary, FTW field-confidence lesson, Kenya polygons, and FTW provenance.

**Start from the chapter evidence:** [AR6 grid lesson](15_ar6_grid_maps.ipynb) · [grid summary](https://github.com/jltobias/JupyterLite-Learning-Atlas/blob/main/content/courses/zarr-sql-views/data/ar6-grid-summary.json) (`courses/zarr-sql-views/data/ar6-grid-summary.json`) · [FTW field-confidence lesson](24_ftw_fields_confidence.ipynb) · [Kenya polygons](https://github.com/jltobias/JupyterLite-Learning-Atlas/blob/main/content/courses/zarr-sql-views/data/ftw_kenya_fields.geojson) (`courses/zarr-sql-views/data/ftw_kenya_fields.geojson`) · [FTW provenance](https://github.com/jltobias/JupyterLite-Learning-Atlas/blob/main/content/courses/zarr-sql-views/data/ftw-provenance.json) (`courses/zarr-sql-views/data/ftw-provenance.json`)

- **[RustGIS Web](https://web.rustgis.app/):** Select small existing FTW polygons with provenance; compare only actual supported confidence fields. Keep synthetic cube-support accounting separate. In RustGIS, import the reviewed geographic export through Add Data; inspect the selected feature IDs and attribute table before styling.
- **[GeoLibre Web](https://web.geolibre.app/):** Select small existing FTW polygons with provenance; compare only actual supported confidence fields. Keep synthetic cube-support accounting separate. In GeoLibre, import that same reviewed file, compare a second legend/filter, and reconcile its count and one known attribute with the notebook.

**Handoff:** Use the [GIS workbench](../../gis-workbench.ipynb) to validate/export an attributed WGS84 layer. Download the GeoJSON, method record and review prompt from `exports/`, then import through the chosen app’s file picker. Compare `_atlas_source_id` (or source `GEOID`) with the original IDs; app-generated internal IDs may differ. Return changed results by explicit upload. Apps and the notebook share no files or kernel state. Practice-only Census data remain Fresno 2020, never this subject’s observations.

**Optional AI prompt (approved subject evidence only):** “Separate predicted field, ownership, embedding, scenario and hazard. Cite supplied feature IDs/fields, separate observations from assumptions, abstain when evidence is missing, and propose one falsifiable check.” For the common practice package, use its generated schema/summary prompt instead. A human-written answer completes the same task.

**GeoAI opportunity:** Embeddings/classification/cube inference are source continuations. Any later fitted exercise must name target/features and spatial/temporal support, reserve a spatial/temporal holdout, inspect leakage and uncertainty, and state evidence that would disprove the interpretation. Styling, thresholds, schema checks and chat alone are not model fitting.

**Human evidence check:** Preserve units and variable-specific support; no copied GPL binaries or claim helper runs cubes. Ask the intended reader which decision the map supports; change one legend, filter or annotation, then recheck interpretation against the same IDs and values.
