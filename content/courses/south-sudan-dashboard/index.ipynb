## Browser GIS pathway

**Geographic question:** Which missing input blocks a credible continuity map?

**Data readiness (O + C):** original staffed-slots/access exercise and source dashboard notebook. Its required facility CSV, HSTP JSON, DEM overlay/bounds and H3 layers are absent locally. Its ACLED/IDP sample generators do not make the missing facility inputs available.

**Start from the chapter evidence:** [original staffed-slots/access exercise](../../projects/south-sudan-dashboard/index.ipynb) · [source dashboard notebook](SouthSudanDashboard_JupyterLite.ipynb)

- **[RustGIS Web](https://web.rustgis.app/):** Inventory absent facility/HSTP/DEM/H3 inputs. Practice only unless a learner supplies an approved attributed geographic layer with actual capacity/access fields. In RustGIS, import the reviewed geographic export through Add Data; inspect the selected feature IDs and attribute table before styling.
- **[GeoLibre Web](https://web.geolibre.app/):** Inventory absent facility/HSTP/DEM/H3 inputs. Practice only unless a learner supplies an approved attributed geographic layer with actual capacity/access fields. In GeoLibre, import that same reviewed file, compare a second legend/filter, and reconcile its count and one known attribute with the notebook.

**Handoff:** Use the [GIS workbench](../../gis-workbench.ipynb) to validate/export an attributed WGS84 layer. Download the GeoJSON, method record and review prompt from `exports/`, then import through the chosen app’s file picker. Compare `_atlas_source_id` (or source `GEOID`) with the original IDs; app-generated internal IDs may differ. Return changed results by explicit upload. Apps and the notebook share no files or kernel state. Practice-only Census data remain Fresno 2020, never this subject’s observations.

**Optional AI prompt (approved subject evidence only):** “Separate capacity, physical access and missing evidence. Cite supplied feature IDs/fields, separate observations from assumptions, abstain when evidence is missing, and propose one falsifiable check.” For the common practice package, use its generated schema/summary prompt instead. A human-written answer completes the same task.

**GeoAI opportunity:** Real access modelling needs verified roads/times/facilities. Any later fitted exercise must name target/features and spatial/temporal support, reserve a spatial/temporal holdout, inspect leakage and uncertainty, and state evidence that would disprove the interpretation. Styling, thresholds, schema checks and chat alone are not model fitting.

**Human evidence check:** No real facility/conflict/displacement status is fabricated. Ask the intended reader which decision the map supports; change one legend, filter or annotation, then recheck interpretation against the same IDs and values.
