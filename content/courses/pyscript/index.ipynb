## Browser GIS pathway

**Geographic question:** Do two runtimes preserve the same portable result?

**Data readiness (B):** data exploration, 24 simulated NYC observations, app-authoring lesson, and round-trip companion.

**Start from the chapter evidence:** [data exploration](02-explore-data.ipynb) · [24 simulated NYC observations](https://github.com/jltobias/JupyterLite-Learning-Atlas/blob/main/content/courses/pyscript/data/nyc-observations.geojson) (`courses/pyscript/data/nyc-observations.geojson`) · [app-authoring lesson](03-author-an-app.ipynb) · [round-trip companion](../../projects/pyscript/index.ipynb)

- **[RustGIS Web](https://web.rustgis.app/):** Select an hour from existing simulated NYC point GeoJSON; compare round-trip IDs/count/values after import. In RustGIS, import the reviewed geographic export through Add Data; inspect the selected feature IDs and attribute table before styling.
- **[GeoLibre Web](https://web.geolibre.app/):** Select an hour from existing simulated NYC point GeoJSON; compare round-trip IDs/count/values after import. In GeoLibre, import that same reviewed file, compare a second legend/filter, and reconcile its count and one known attribute with the notebook.

**Handoff:** Use the [GIS workbench](../../gis-workbench.ipynb) to validate/export an attributed WGS84 layer. Download the GeoJSON, method record and review prompt from `exports/`, then import through the chosen app’s file picker. Compare `_atlas_source_id` (or source `GEOID`) with the original IDs; app-generated internal IDs may differ. Return changed results by explicit upload. Apps and the notebook share no files or kernel state. Practice-only Census data remain Fresno 2020, never this subject’s observations.

**Optional AI prompt (approved subject evidence only):** “Diagnose a swapped coordinate or incompatible unit using supplied rows. Cite supplied feature IDs/fields, separate observations from assumptions, abstain when evidence is missing, and propose one falsifiable check.” For the common practice package, use its generated schema/summary prompt instead. A human-written answer completes the same task.

**GeoAI opportunity:** Runtime agreement is not model validation. Any later fitted exercise must name target/features and spatial/temporal support, reserve a spatial/temporal holdout, inspect leakage and uncertainty, and state evidence that would disprove the interpretation. Styling, thresholds, schema checks and chat alone are not model fitting.

**Human evidence check:** Keep simulation labels; wait-time companion stays separate without ID/unit crosswalk. Ask the intended reader which decision the map supports; change one legend, filter or annotation, then recheck interpretation against the same IDs and values.
