## Browser GIS pathway

**Geographic question:** What connects a flow count to endpoints and time?

**Data readiness (O + C):** flow lesson, inventory, and original A–D flow exercise.

**Start from the chapter evidence:** [flow lesson](contents/notebooks/02_mobility_flowmap.ipynb) · [inventory](contents/notebooks/01_data_inventory.ipynb) · [original A–D flow exercise](../../projects/drc-population-mobility-border-mapping/index.ipynb)

- **[RustGIS Web](https://web.rustgis.app/):** Inventory remote/conditional inputs. Keep A-D flows nonspatial; geographic flow work needs a learner-approved attributed export with explicit endpoint coordinates/dates. In RustGIS, import the reviewed geographic export through Add Data; inspect the selected feature IDs and attribute table before styling.
- **[GeoLibre Web](https://web.geolibre.app/):** Inventory remote/conditional inputs. Keep A-D flows nonspatial; geographic flow work needs a learner-approved attributed export with explicit endpoint coordinates/dates. In GeoLibre, import that same reviewed file, compare a second legend/filter, and reconcile its count and one known attribute with the notebook.

**Handoff:** Use the [GIS workbench](../../gis-workbench.ipynb) to validate/export an attributed WGS84 layer. Download the GeoJSON, method record and review prompt from `exports/`, then import through the chosen app’s file picker. Compare `_atlas_source_id` (or source `GEOID`) with the original IDs; app-generated internal IDs may differ. Return changed results by explicit upload. Apps and the notebook share no files or kernel state. Practice-only Census data remain Fresno 2020, never this subject’s observations.

**Optional AI prompt (approved subject evidence only):** “Explain why connectivity does not establish infection/transmission. Cite supplied feature IDs/fields, separate observations from assumptions, abstain when evidence is missing, and propose one falsifiable check.” For the common practice package, use its generated schema/summary prompt instead. A human-written answer completes the same task.

**GeoAI opportunity:** Mobility/access models require dated flows, roads and denominators. Any later fitted exercise must name target/features and spatial/temporal support, reserve a spatial/temporal holdout, inspect leakage and uncertainty, and state evidence that would disprove the interpretation. Styling, thresholds, schema checks and chat alone are not model fitting.

**Human evidence check:** No guessed geocoding, invented DRC flows or degrees-as-km catchments. Ask the intended reader which decision the map supports; change one legend, filter or annotation, then recheck interpretation against the same IDs and values.
