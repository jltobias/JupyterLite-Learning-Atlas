## Browser GIS pathway

**Geographic question:** Is largest drop the same as steepest descent?

**Data readiness (O + C):** orthogonal descent companion, hydrology lesson, architecture boundary, and dataset notes.

**Start from the chapter evidence:** [orthogonal descent companion](../../projects/whitebox/index.ipynb) · [hydrology lesson](04_Hydrological_Analysis.ipynb) · [architecture boundary](01_Architecture_and_Compatibility.ipynb) · [dataset notes](https://github.com/jltobias/JupyterLite-Learning-Atlas/blob/main/content/courses/whitebox/datasets.md) (`courses/whitebox/datasets.md`)

- **[RustGIS Web](https://web.rustgis.app/):** Preserve 3x3 planar descent. Practice only with an attributed terrain summary if supplied; otherwise shared Census mechanics. In RustGIS, import the reviewed geographic export through Add Data; inspect the selected feature IDs and attribute table before styling.
- **[GeoLibre Web](https://web.geolibre.app/):** Preserve 3x3 planar descent. Practice only with an attributed terrain summary if supplied; otherwise shared Census mechanics. In GeoLibre, import that same reviewed file, compare a second legend/filter, and reconcile its count and one known attribute with the notebook.

**Handoff:** Use the [GIS workbench](../../gis-workbench.ipynb) to validate/export an attributed WGS84 layer. Download the GeoJSON, method record and review prompt from `exports/`, then import through the chosen app’s file picker. Compare `_atlas_source_id` (or source `GEOID`) with the original IDs; app-generated internal IDs may differ. Return changed results by explicit upload. Apps and the notebook share no files or kernel state. Practice-only Census data remain Fresno 2020, never this subject’s observations.

**Optional AI prompt (approved subject evidence only):** “Explain diagonal-neighbor and unequal-spacing effects. Cite supplied feature IDs/fields, separate observations from assumptions, abstain when evidence is missing, and propose one falsifiable check.” For the common practice package, use its generated schema/summary prompt instead. A human-written answer completes the same task.

**GeoAI opportunity:** Native WhiteboxTools/raster workflows remain continuations. Any later fitted exercise must name target/features and spatial/temporal support, reserve a spatial/temporal holdout, inspect leakage and uncertainty, and state evidence that would disprove the interpretation. Styling, thresholds, schema checks and chat alone are not model fitting.

**Human evidence check:** No invented terrain geography, routed stream or native Pyodide executable. Ask the intended reader which decision the map supports; change one legend, filter or annotation, then recheck interpretation against the same IDs and values.
