## Browser GIS pathway

**Geographic question:** Is a severe signal useful when stale or incomplete?

**Data readiness (B):** 2D lesson, training districts, surveillance, facility table, and provenance.

**Start from the chapter evidence:** [2D lesson](03_2D_Maps.ipynb) · [training districts](https://github.com/jltobias/JupyterLite-Learning-Atlas/blob/main/content/courses/world-monitor-prototype/data/districts.geojson) (`courses/world-monitor-prototype/data/districts.geojson`) · [surveillance](https://github.com/jltobias/JupyterLite-Learning-Atlas/blob/main/content/courses/world-monitor-prototype/data/surveillance.csv) (`courses/world-monitor-prototype/data/surveillance.csv`) · [facility table](https://github.com/jltobias/JupyterLite-Learning-Atlas/blob/main/content/courses/world-monitor-prototype/data/facilities.csv) (`courses/world-monitor-prototype/data/facilities.csv`) · [provenance](https://github.com/jltobias/JupyterLite-Learning-Atlas/blob/main/content/courses/world-monitor-prototype/data/provenance.json) (`courses/world-monitor-prototype/data/provenance.json`)

- **[RustGIS Web](https://web.rustgis.app/):** Join a fixed date of bundled surveillance to training districts; preserve reporting completeness, denominators and data age. In RustGIS, import the reviewed geographic export through Add Data; inspect the selected feature IDs and attribute table before styling.
- **[GeoLibre Web](https://web.geolibre.app/):** Join a fixed date of bundled surveillance to training districts; preserve reporting completeness, denominators and data age. In GeoLibre, import that same reviewed file, compare a second legend/filter, and reconcile its count and one known attribute with the notebook.

**Handoff:** Use the [GIS workbench](../../gis-workbench.ipynb) to validate/export an attributed WGS84 layer. Download the GeoJSON, method record and review prompt from `exports/`, then import through the chosen app’s file picker. Compare `_atlas_source_id` (or source `GEOID`) with the original IDs; app-generated internal IDs may differ. Return changed results by explicit upload. Apps and the notebook share no files or kernel state. Practice-only Census data remain Fresno 2020, never this subject’s observations.

**Optional AI prompt (approved subject evidence only):** “Draft three ID/date-cited briefing claims with missingness limits. Cite supplied feature IDs/fields, separate observations from assumptions, abstain when evidence is missing, and propose one falsifiable check.” For the common practice package, use its generated schema/summary prompt instead. A human-written answer completes the same task.

**GeoAI opportunity:** Optional anomaly detection needs evaluation; freshness weighting is heuristic. Any later fitted exercise must name target/features and spatial/temporal support, reserve a spatial/temporal holdout, inspect leakage and uncertainty, and state evidence that would disprove the interpretation. Styling, thresholds, schema checks and chat alone are not model fitting.

**Human evidence check:** No live-monitor claim; counts/rates/completeness remain distinct; depth retained for 2D export. Ask the intended reader which decision the map supports; change one legend, filter or annotation, then recheck interpretation against the same IDs and values.
